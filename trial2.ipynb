```python
import numpy as np
import csv

tabel = np.zeros((2, 3))   # 2 bulan, 3 produk, semua mulai dari 0

with open(r"C:\Users\sayyi\Downloads\Zomato Restaurant Dataset.csv") as f:
    pembaca = csv.reader(f)
    next(pembaca)                       # lewati baris header
    for baris in pembaca:
        bulan = int(baris[0])
        produk = int(baris[1])
        jumlah = float(baris[2])
        tabel[bulan, produk] = jumlah

print(tabel)
```
