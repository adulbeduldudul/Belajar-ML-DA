````python
import pandas as pd
import numpy as np

# pengaturan tampilan biar bersih
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 1000)

path=r'C:\Users\sayyi\Downloads\sales.csv'

null_value = ['None', 'none', 'null', '?', ' ', '   ', 'n/a', 'na', '--', '-']

csv = pd.read_csv(path, na_values=null_value)

# output
print("Jumlah Data: ",csv.shape)
print("\nKolom & tipe data: \n", csv.dtypes)
print("\n5 baris pertama:")
csv.head()


````

deteksi Nan value menggunakan metode eksplisit


````python
count_null=csv.isna().sum()
percent_null=(csv.isna().mean()*100).round(2)

summary=pd.DataFrame({ 
    'Missing':count_null,
    'Percentage':percent_null
}).sort_values('Missing',ascending=False)

display(summary[summary['Missing']>0])
````

````python
pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
pd.set_option('display.width',100)
cat_col=['Gender','City','Region','Sales_Channel','Payment_Method', 'Customer_Segment', 'Product_Category', 'Order_Status', 'Return_Flag', 'Marketing_Channel']

for col in cat_col:
    print(f'\n=== {col} ({csv[col].nunique()} variasi unik) ===')
    print(csv[col].value_counts(dropna=False))
    print()
````

standarisasi nilai yang tidak konsisten

````python
csv['Gender']=csv['Gender'].str.lower().str.strip()
Gender_map={
    'male': 'Male',
    'm': 'Male',
    'female': 'Female',
    'f': 'Female',
    'non-binary': 'Non-binary',
    'nan': 'N/A',
    'prefer not to say': 'Prefer not to say'
}
csv['Gender']=csv['Gender'].map(Gender_map)
print(csv['Gender'].value_counts(dropna=False))
````

````python
print(csv['Gender'].unique().tolist())
````

````python
csv['City'] = csv['City'].str.lower().str.strip()

city_map = {
    'karachi': 'Karachi',
    'lahore': 'Lahore',
    'rawalpindi': 'Rawalpindi',
    'islamabad': 'Islamabad',
    'faisalabad': 'Faisalabad',
    'peshawar': 'Peshawar',
    'multan': 'Multan',
    'gujranwala': 'Gujranwala',
    'hyderabad': 'Hyderabad',
    'sialkot': 'Sialkot',
    'quetta': 'Quetta'
}

csv['City'] = csv['City'].map(city_map)

print(csv['City'].value_counts(dropna=False))
````

````python
csv['Sales_Channel']=csv['Sales_Channel'].str.lower().str.strip()

Sales_Channel_map={
    'online': 'Online',
    'e-commerce': 'Online',       # E-commerce = Online
    'marketplace': 'Marketplace',
    'retail store': 'Retail Store',
    'retail': 'Retail Store',      # Retail = Retail Store
    'phone order': 'Phone Order'
}
csv['Sales_Channel']=csv['Sales_Channel'].map(Sales_Channel_map)
print(csv['Sales_Channel'].value_counts(dropna=False))
````

````python
csv['Payment_Method']=csv['Payment_Method'].str.lower().str.strip()

Payment_Method_map={
'COD':'Cash On Delivery',
'cash on delivery' : 'Cash On Delivery',
'CASH ON DELIVERY':'Cash On Delivery',
'Wallet':'Digital Wallet',
 'DIGITAL WALLET' : 'Digital Wallet',
'digital wallet': 'Digital Wallet',
'debit card': 'Debit Card',
'credit card': 'Credit Card',
'bank tf':'Bank Transfer'
}

csv['Payment_Method']=csv['Payment_Method'].map(Payment_Method_map)
print(csv['Payment_Method'].value_counts(dropna=False))
````
