````python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
pd.set_option('display.width',1000)

path=r"C:\Users\sayyi\Downloads\sales_dataset_89984_records_23_features.csv"
missing_format=['None', 'none', 'null', '?', ' ', '   ', 'n/a', 'na', '--', '-']
csv=pd.read_csv(path, na_values=missing_format)

print(csv.shape)
print(csv.dtypes)
csv.head(4)
````

````python
null_count=csv.isnull().sum()
null_percent=(csv.isnull().mean()*100).round(2)
summary=pd.DataFrame({
    'Null Count':null_count,
    'Null Percent':null_percent
}).sort_values('Null Count',ascending=False)
summary
````
